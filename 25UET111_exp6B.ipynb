```python
student={"Name":"ABC",
         "Age":"20",
         "branch":"entc"}
print(student)
```

```python
# add elememt--
student={"Name":"ABC",
         "Age":20,
         "branch":"entc"}
student["cgpa"]=9.5
print(student)
```

```python
# change/update value--
student={"name":"ABC",
         "age":"20",
         "branch":"entc"}
student["age"]=25
print(student)
```

```python
# update()-- to add multiple items
student={"Name":"ABC",
         "Age":"20",
         "branch":"entc"}
student.update({
    "sem":3,
    "result": "pass"
})
print(student)
```

```python
# access value by key--
student={"name":"ABC",
         "age":"20",
         "branch":"entc"}
print(student["name"])
```

```python
# access value by get90
student={"name":"ABC",
         "age":"20",
         "branch":"entc"}
print(student.get("sem"))
```

```python
# keys()
student={"name":"ABC",
         "age":"20",
         "branch":"entc"}
print(student.keys())
```

```python
# values()
student={"name":"ABC",
         "age":"20",
         "branch":"entc"}
print(student.values())
```

```python
# itms90
student={"name":"ABC",
         "age":"20",
         "branch":"entc"}
print(student.items())
```

```python
student={"name":"ABC",
         "age":"20",
         "branch":"entc"}
for key, value in student.items():
    print(key, "=", value)
```

```python
# pop()
student={"name":"ABC",
         "age":"20",
         "branch":"entc"}
student.pop("age")
print(student)
```

```python
# popitem-- removes last item
student={"name":"ABC",
         "age":"20",
         "branch":"entc"}
student.popitem()
print(student)
```

```python
# nested dictionary--
student={"student1":
        {
            "name":"abc",
            "age":20,
            "branch": "entc"
        },
         "student2":
         {
              "name":"xyz",
            "age":21,
            "branch": "ele"
         },
         "student3":
         {
              "name":"pqr",
            "age":22,
            "branch": "civil"
         }
         
        }
print(student)
```
